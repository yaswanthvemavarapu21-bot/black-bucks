```python
%%writefile data_processor.py
# data_processor.py - Data Acquisition & Technical Feature Engineering
import yfinance as yf
import pandas as pd
import numpy as np

def fetch_stock_data(ticker: str, start_date: str, end_date: str) -> pd.DataFrame:
    """Fetches historical daily OHLCV equity data from Yahoo Finance API."""
    df = yf.download(ticker, start=start_date, end=end_date)
    df.dropna(inplace=True)
    return df

def calculate_technical_indicators(df: pd.DataFrame) -> pd.DataFrame:
    """Calculates EMA, RSI, MACD, and target signals."""
    # Exponential Moving Averages (EMA 20 & EMA 50)
    df['EMA_20'] = df['Close'].ewm(span=20, adjust=False).mean()
    df['EMA_50'] = df['Close'].ewm(span=50, adjust=False).mean()

    # Relative Strength Index (RSI 14)
    delta = df['Close'].diff()
    gain = (delta.where(delta > 0, 0)).rolling(window=14).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(window=14).mean()
    rs = gain / (loss + 1e-8)
    df['RSI_14'] = 100 - (100 / (1 + rs))

    # MACD & Signal Line
    ema_12 = df['Close'].ewm(span=12, adjust=False).mean()
    ema_26 = df['Close'].ewm(span=26, adjust=False).mean()
    df['MACD'] = ema_12 - ema_26
    df['MACD_Signal'] = df['MACD'].ewm(span=9, adjust=False).mean()

    # Target Direction (1 if Close(t+1) > Close(t) else 0)
    df['Target_Direction'] = (df['Close'].shift(-1) > df['Close']).astype(int)
    df.fillna(0, inplace=True)
    return df
```

```python
%%writefile sentiment_nlp.py
# sentiment_nlp.py - FinBERT News Sentiment Analysis Pipeline
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
import numpy as np

class FinancialSentimentAnalyzer:
    def __init__(self):
        self.tokenizer = AutoTokenizer.from_pretrained("ProsusAI/finbert")
        self.model = AutoModelForSequenceClassification.from_pretrained("ProsusAI/finbert")

    def analyze_headlines(self, headlines: list) -> float:
        """Computes continuous aggregate polarity sentiment score for daily headlines."""
        if not headlines:
            return 0.0

        inputs = self.tokenizer(headlines, padding=True, truncation=True, return_tensors="pt")
        outputs = self.model(**inputs)
        predictions = torch.nn.functional.softmax(outputs.logits, dim=-1)

        # Polarity Weights: Positive=1.0, Negative=-1.0, Neutral=0.0
        weights = torch.tensor([1.0, -1.0, 0.0])
        scores = torch.matmul(predictions, weights)
        return float(torch.mean(scores).detach().numpy())
```

```python
%%writefile model_training.py
# model_training.py - Hybrid LSTM-GRU Predictive Deep Learning Model
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, GRU, Dense, Dropout

def create_sliding_windows(data: np.ndarray, seq_length: int = 60):
    """Formats 2D feature matrices into sequential tensors for recurrent networks."""
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i : i + seq_length, :-1])
        y.append(data[i + seq_length, -1])
    return np.array(X), np.array(y)

def build_hybrid_lstm_model(input_shape: tuple) -> Sequential:
    """Builds stacked LSTM-GRU architecture for trend classification."""
    model = Sequential([
        LSTM(128, return_sequences=True, input_shape=input_shape),
        Dropout(0.2),
        GRU(64, return_sequences=False),
        Dropout(0.2),
        Dense(32, activation='relu'),
        Dense(1, activation='sigmoid')  # Output binary directional probability
    ])
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    return model
```

```python
%%writefile app.py
# app.py - Streamlit User Interface & Dashboard Logic
import streamlit as st
import pandas as pd
import plotly.graph_objects as go
from data_processor import fetch_stock_data, calculate_technical_indicators

st.set_page_config(page_title="Stock Market Trend Dashboard", layout="wide")
st.title("📈 Stock Market Trend Analysis & AI Forecasting Engine")

ticker = st.sidebar.text_input("Select Stock Ticker Symbol", value="AAPL")
start_date = st.sidebar.date_input("Start Date", pd.to_datetime("2020-01-01"))
end_date = st.sidebar.date_input("End Date", pd.to_datetime("today"))

if st.sidebar.button("Execute Market Analysis"):
    df = fetch_stock_data(ticker, str(start_date), str(end_date))
    df = calculate_technical_indicators(df)

    # Plot Candlestick Chart with Technical Overlays
    fig = go.Figure()
    fig.add_trace(go.Candlestick(x=df.index, open=df['Open'], high=df['High'],
                                 low=df['Low'], close=df['Close'], name='OHLC'))
    fig.add_trace(go.Scatter(x=df.index, y=df['EMA_20'], line=dict(color='orange'), name='20-Day EMA'))
    fig.add_trace(go.Scatter(x=df.index, y=df['EMA_50'], line=dict(color='blue'), name='50-Day EMA'))

    st.plotly_chart(fig, use_container_width=True)
    st.success("Technical Features and Interactive Chart Generated Successfully!")
```
